```python
import os
import sys
sys.path.append(os.path.abspath("."))
from src.meamt_demo_wrapper import MEAMT
import moeabench as mb
```

```python
exp = mb.experiment()

exp.mop = mb.mops.DTLZ8()
alg = MEAMT(population=400, generations=1000, tabelas="todas", incluir_completa=False)
exp.moea = alg
exp.run()
```

```python
info = alg.info

import matplotlib.pyplot as plt

ax = plt.figure(figsize=(7, 7)).add_subplot(projection="3d")
for nome in info["nomes"]:
    F = info["tabelas"][nome]["F"]
    if nome != "{f1, f2, f3}":
        ax.scatter(F[:, 0], F[:, 1], F[:, 2], s=15, label=nome)
        
ax.set_xbound(0, 1); ax.set_ybound(0, 1); ax.set_zbound(0, 1)
ax.set_xlabel("f1"); ax.set_ylabel("f2"); ax.set_zlabel("f3")
ax.legend()
plt.show()
```
